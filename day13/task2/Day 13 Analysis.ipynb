# Ecommerce Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ecommerce_dirty_practice_20.csv")
df.head()
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors="coerce", dayfirst=True)
df["Month"] = df["OrderDate"].dt.to_period("M").astype("string")

df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Revenue"] = df["Quantity"] * df["UnitPrice"] * (1 - df["DiscountPercent"] / 100)

df.head()
```

```python
df[["UnitPrice", "Quantity", "DiscountPercent", "Rating", "Revenue"]].describe()
```

```python
df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)
```

```python
df.groupby("Product")["Quantity"].sum().sort_values(ascending=True)
```

```python
product_summary = df.groupby("Product").agg(
    TotalRevenue=("Revenue", "sum"),
    UnitsSold=("Quantity", "sum"),
    Orders=("OrderID", "count")
).sort_values("TotalRevenue", ascending=False)
product_summary
```

```python
city_summary = df.groupby("City").agg(
    TotalRevenue=("Revenue", "sum"),
    Orders=("OrderID", "count"),
    AverageRevenue=("Revenue", "mean")
).sort_values("TotalRevenue", ascending=False)
city_summary
```

```python
monthly_summary = df.groupby("Month").agg(
    TotalRevenue=("Revenue", "sum"),
    Orders=("OrderID", "count")
).sort_index()
monthly_summary
```

```python
df.groupby("PaymentMethod").agg(
    TotalRevenue=("Revenue", "sum"),
    UnitsSold=("Quantity", "sum"),
    Orders=("OrderID", "count")
)
```

```python
df.groupby(["City", "Product"]).agg(
    UnitsSold=("Quantity", "sum"),
    TotalRevenue=("Revenue", "sum")
)
```

```python
print("Total units sold:", df["Quantity"].sum())
print("Total revenue:", round(df["Revenue"].sum(), 2))
print("Average order revenue:", round(df["Revenue"].mean(), 2))
print("Highest order revenue:", round(df["Revenue"].max(), 2))
print("Lowest order revenue:", round(df["Revenue"].min(), 2))
```

```python
product_summary["TotalRevenue"].plot(kind="bar", title="Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_summary["TotalRevenue"].plot(kind="line", marker="o", title="Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
city_summary["TotalRevenue"].plot(kind="bar", title="Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()
```

```python
df.plot(kind="scatter", x="Quantity", y="Revenue", title="Quantity vs Revenue")
plt.tight_layout()
plt.show()
```

```python
df.to_csv("ecommerce_analysis_output.csv", index=False)
product_summary.to_csv("ecommerce_product_summary.csv")
monthly_summary.to_csv("ecommerce_monthly_summary.csv")
```
